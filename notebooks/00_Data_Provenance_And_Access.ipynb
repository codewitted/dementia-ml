# 00 - Data Provenance and Access

Distinction-level documentation on the provenance, access, and practical challenges associated with core data sources: **ADNI** and **OASIS-1**.


## 1. Data Sources Overview

### Alzheimer's Disease Neuroimaging Initiative (ADNI)
- [ADNI Website](https://adni.loni.usc.edu/)
- ADNI is a longitudinal multicenter study designed to develop clinical, imaging, genetic, and biochemical biomarkers for the early detection and tracking of Alzheimer's disease.
- Includes MRI, PET, genetics, cognitive tests, CSF, and clinical data.

### OASIS-1 (Open Access Series of Imaging Studies)
- [OASIS Brains](https://www.oasis-brains.org/)
- Cross-sectional MRI data for a cohort including both healthy aging and dementia cases.
- Includes anonymized T1-weighted MRI scans and demographic/cognitive measures.


## 2. Access and Download

**ADNI**:
- Requires registration and data use agreement approval. Approval is **not instant** (may take days).
- After approval, downloads are performed via the LONI portal.
- Data is split into many zipped segments.
- Practical challenge: Large data size (>100GB), fragmented downloads, unstable sessions.

**OASIS-1**:
- Freely available for download, but requires agreeing to terms of use.
- Download is straightforward from the OASIS brains website.
- Practical challenge: Dataset is smaller and lacks longitudinal detail compared to ADNI.

**General Obstacles**:
- For both datasets, raw DICOM/NIfTI imaging formats require substantial preprocessing.
- Variability in acquisition protocols (scanner differences, year of acquisition, etc.).
    

## 3. Pivots and Decisions

- **Which dataset to use for what task:**
  - ADNI: Used for modeling longitudinal progression, and for multi-modal machine learning.
  - OASIS-1: Used for cross-sectional benchmarking or simpler proof-of-concept pipelines.
- **Subsetting participants:**
  - Due to size and download restrictions, practical experiments performed on a subset of the ADNI sample.
  - Pivoted to OASIS-1 for rapid prototyping due to easier access.

- **Data transformation:**
  - Intensive preprocessing (skull-stripping, spatial normalization, etc.) required for MRI images.
  - Use of standard neuroimaging processing pipelines for harmonization.
    

## 4. Practical Obstacles and Mitigations

- Approval wait times for ADNI increased project latency.
- Download limits and connection interruptions necessitated the use of download managers or CLI tools.
- High computational requirement for image preprocessing required access to cloud or high-performance local infrastructure.
- Mapping clinical/diagnostic labels across datasets needed substantial data cleaning given different coding schemes.
